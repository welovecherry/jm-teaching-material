## 1. 변수와 자료형 · 55–80분

### 📖 왜 '변수'부터 배울까?

1. **프로그램은 결국 값을 다루는 일**입니다. 로그인 실패 횟수, 사용자 이름, 경보 여부 — 전부 '값'입니다.
2. 그 값을 어딘가 **담아 두어야** 나중에 꺼내 쓰고, 비교하고, 계산합니다. 그 담는 그릇이 **변수**입니다.
3. 오늘 배우는 나머지(자료형·리스트·딕셔너리)도 결국 **"변수에 무엇을, 어떻게 담느냐"**의 이야기입니다.
4. **이 과정과의 연결**: 앞으로 만들 보안 자동화 봇은 "로그를 **변수에 담고** → 조건으로 걸러내고 → 알림 보내기"의 반복입니다.

> 🍳 **비유:** 변수는 재료를 담는 **그릇**입니다. 그릇이 있어야 요리를 시작할 수 있습니다.

### 📖 변수 = 값에 붙이는 '이름표'

- 변수는 `이름 = 값` 으로 만듭니다. 예: `x = 7` → "값 7에 `x`라는 이름표를 붙임".
- 한 번 만들면, 그 **이름으로 값을 꺼내** 씁니다.
- 값은 나중에 바꿀 수 있습니다(재할당). 같은 이름에 새 값을 넣으면 됩니다.

> **주의 — `=` 는 "같다"가 아닙니다.**
> - `=` : "값을 **넣어라**"(대입)
> - `==` : "**같은지** 물어봐라"(비교) — 뒤 3번에서 다룸

````python
# 💻 예제 — 가장 간단한 변수 하나
name = "kim"        # "kim" 이라는 값에 name 이라는 이름표를 붙임
print(name)         # 이름표로 값을 꺼내 출력
````

**✅ 성공 출력**
```
kim
```

````python
# 💻 예제 — 값을 바꾸기(재할당)
count = 3
print(count)        # 3
count = 10          # 같은 이름에 새 값을 넣음
print(count)        # 10
````

**✅ 성공 출력**
```
3
10
```

````python
# ✍️ 문제 (출력 예측) — 아래가 무엇을 출력할지 먼저 예측한 뒤 실행하세요.
city = "seoul"
city = "busan"
print(city)         # 예측: ?
````

### 📖 값에는 '종류'가 있습니다 = 자료형

사람도 "숫자 7"과 "글자 ㄱ"을 다르게 다루듯, 컴퓨터도 값의 **종류**를 구분합니다. 기본 4가지입니다.

1. **문자열 `str`** — 글자(문자)들의 나열. **반드시 따옴표로 감쌉니다.** 예: `"kim01"`, `"login_failed"`
   - 따옴표가 "이건 글자야"라고 컴퓨터에 알려주는 표시입니다. 작은따옴표 `'...'`, 큰따옴표 `"..."` 둘 다 됩니다.
2. **정수 `int`** — 소수점 없는 숫자. 따옴표 없이 씁니다. 예: `7`, `0`, `-3`
3. **실수 `float`** — 소수점 있는 숫자. 예: `73.5`, `0.23`
4. **참거짓 `bool`** — `True`(참) 또는 `False`(거짓) 둘 중 하나. 조건 판단에 씁니다.

- 값의 종류는 `type(값)` 으로 확인합니다.

> **왜 종류가 중요할까?** 같은 7이라도 **글자 `"7"`** 과 **숫자 `7`** 은 컴퓨터에게 완전히 다릅니다.
> (이 차이는 3번 '형변환'에서 실감하게 됩니다.)

````python
# 💻 예제 — 문자열(str): 글자는 따옴표로 감싼다
user = "kim01"
event = "login_failed"
print(user)
print(event)
````

**✅ 성공 출력**

````python
# 💻 예제 — 숫자: 정수(int)와 실수(float)
failed_login_count = 7      # 정수
cpu_usage = 73.5            # 실수(소수점)
print(failed_login_count)
print(cpu_usage)
````

**✅ 성공 출력**

````python
# 💻 예제 — 참거짓(bool): True 또는 False
is_alert = True
is_blocked = False
print(is_alert)
print(is_blocked)
````

**✅ 성공 출력**

````python
# 💻 예제 — 값의 '종류'를 type() 으로 확인
print(type("kim01"))   # str
print(type(7))         # int
print(type(73.5))      # float
print(type(True))      # bool
````

### 📖 변수 이름 잘 짓는 법

이름에는 **규칙(어기면 에러)** 과 **관습(어겨도 실행은 되지만 지키는 게 좋음)** 이 있습니다.

**규칙 — 어기면 에러가 납니다**
1. 영문·숫자·밑줄(`_`)만 사용
2. **숫자로 시작 금지** (`2fail` ✗)
3. 공백·하이픈 금지 (`fail count` ✗, `fail-count` ✗)
4. 예약어 금지 (`if`, `for`, `class` 등은 이름으로 못 씀)

**관습 — 팀·실무의 약속 (안 지켜도 실행은 됨)**
1. 파이썬 표준은 **`snake_case`** — 소문자 단어를 밑줄로 연결. 예: `failed_login_count`
2. **`is_` / `has_` / `can_`** 로 시작하면 보통 **참거짓(bool)** 값을 뜻함
3. **복수형**(`users`, `logs`)은 보통 **여러 개(리스트)** 를 뜻함
4. **안 바뀌는 값(상수)** 은 **대문자** 로. 예: `MAX_RETRY = 5`

> **왜 중요할까?** 이름만 봐도 "이건 참거짓이구나, 이건 목록이구나"를 알 수 있어 코드를 빨리 읽고 실수를 줄입니다.

````python
# 💻 예제 — 관습을 지킨 좋은 이름들
failed_login_count = 7          # snake_case + 의미 있는 이름
is_blocked = False              # is_ → 참거짓
users = ["kim01", "lee02"]      # 복수형 → 여러 개(리스트)
MAX_RETRY = 5                   # 대문자 → 상수
print(failed_login_count, is_blocked, users, MAX_RETRY)
````

**✅ 성공 출력**
```
7 False ['kim01', 'lee02'] 5
```

### 📖 f-string — 값을 문장에 끼워 넣기

- 로그 메시지·리포트·알림 문구에 계속 쓰는 핵심 문법입니다. 아주 쉬운 것부터 봅니다.

## 2. 리스트와 딕셔너리 · 80–105분

### 📖 개념

값이 **여러 개**일 때(예: 사용자 여러 명, 로그 여러 줄) 하나씩 변수를 만들면 번거롭습니다.
여러 값을 한 번에 담는 두 가지 그릇이 있습니다.

1. **리스트 `[ ]`** — 값을 **순서대로** 담습니다.
   - 번호(인덱스)로 꺼냅니다. 번호는 **0부터** 셉니다. (`users[0]` = 첫 번째)
   - `len()` 으로 개수를 세고, `append()` 로 뒤에 추가합니다.
2. **딕셔너리 `{ }`** — **이름표(키)-값** 쌍으로 담습니다.
   - `log["user"]` 처럼 **이름으로** 꺼냅니다.
   - 보안 로그 한 줄(user·event·ip)처럼 "이름 있는 정보"에 적합합니다.

````python
# 💻 예제 — 리스트: 여러 사용자
users = ["kim01", "lee02", "park03"]
print(users[0])      # 첫 번째 → kim01
print(len(users))    # 개수 → 3
````

**✅ 성공 출력**
```
kim01
3
```

````python
# 💻 예제 — 딕셔너리: 로그 한 줄
log = {"user": "kim01", "event": "login_failed"}
print(log["user"])     # 이름으로 값 꺼내기 → kim01
print(log["event"])    # → login_failed
````

**✅ 성공 출력**
```
kim01
login_failed
```

````python
# ✍️ 문제 (출력 예측) — 무엇이 나올지 예측한 뒤 실행하세요.
users = ["kim01", "lee02", "park03"]
print(users[1])       # 예측: ?
print(len(users))     # 예측: ?
````

## [심화·자습] 2 — 더 많은 예제와 문제

### 📖 (심화) `.get()` 은 없는 키도 안전합니다
- 대괄호 `log["없는키"]` 는 없는 키를 꺼내면 **에러로 멈춥니다.**
- `log.get("없는키")` 는 없으면 조용히 `None` 을 돌려줍니다(멈추지 않음).
- **집합(set)·튜플(tuple)** 은 뒤 차시에서 배웁니다.

````python
# 💻 (심화) 대괄호 vs .get(), 그리고 append
log = {"user": "kim01"}
print(log.get("country"))   # 없으면 None
users = ["kim01"]
users.append("lee02")       # 뒤에 추가
print(users)
````

**✅ 성공 출력**
```

['kim01', 'lee02']
```

## 3. 연산자와 형변환 · 105–120분

### 📖 개념

**(1) 연산자 — 계산하고 비교하기**
1. 산술: `+  -  *  /` (더하기·빼기·곱하기·나누기)
2. 비교: `>  <  ==  !=` — 결과는 항상 **참거짓(bool)**
3. 논리: `and  or  not` — 여러 조건을 묶기

**(2) 형변환 — 글자와 숫자 바꾸기 (실수 주의)**
- 파일(CSV)에서 읽은 숫자는 사실 **글자**입니다. 예: `"7"`
- 글자끼리는 계산이 안 되므로, `int()` 로 **진짜 숫자로 바꿔** 주어야 합니다.
- `int("7")` → `7` / `str(7)` → `"7"` / `float("3.14")` → `3.14`

> **왜 중요할까?** "글자 숫자를 안 바꿔서" 나는 오류가 실무에 아주 흔합니다.

````python
# 💻 예제 — 연산자 맛보기
print(7 + 3)       # 산술 → 10
print(7 > 5)       # 비교 → True
print(7 > 5 and 3 > 10)   # 논리 → False
````

**✅ 성공 출력**
```
10
True
False
```

````python
# 💻 예제 — 형변환
raw_value = "7"          # 파일에서 읽은 값은 글자
count = int(raw_value)   # 진짜 숫자로 바꿈
print(count + 1)         # 8
````

**✅ 성공 출력**
```
8
```
> `int()` 없이 `"7" + 1` 을 하면 글자+숫자라 **에러(TypeError)** 가 납니다.

````python
# ✍️ 문제 (출력 예측) — 결과를 예측한 뒤 실행하세요.
count = 7
print(count == 7)        # 예측: ?
print(count != 10)       # 예측: ?
````
