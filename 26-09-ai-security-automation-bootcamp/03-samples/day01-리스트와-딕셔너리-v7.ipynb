### 📖 번호(인덱스)로 꺼내기

- 리스트의 값은 **번호(인덱스)** 로 꺼냅니다: `리스트[번호]`
- 번호는 **0부터** 셉니다. 첫 번째 = `[0]`, 두 번째 = `[1]`.

````python
# 💻 예제 — 번호로 값 꺼내기
users = ["kim01", "lee02", "park03"]
print(users[0])   # 첫 번째
print(users[1])   # 두 번째
````

### 📖 잠깐 — '함수'와 '메서드'가 뭔가요?

이제 `len(...)` 과 `append(...)` 가 나옵니다. 처음 보는 모양이죠? 지금은 딱 이만큼만 알면 됩니다.

1. **함수** = 미리 만들어진 명령. **이름 뒤 괄호에 값을 넣어** 어떤 일을 시킵니다.
   예: `len(users)` → 리스트의 개수를 세어 줍니다.
2. **메서드** = **값에 점(`.`)을 찍어 붙는** 함수. 예: `users.append("x")` → 그 리스트에 값을 추가합니다.

> 함수를 **직접 만드는 법**은 Day 2에서 배웁니다. 지금은 "가져다 쓰는 도구"로만 알면 충분합니다.

### 📖 개수 세기(len)와 값 추가(append)

- `len(리스트)` → 값의 **개수** (함수)
- `리스트.append(값)` → 맨 **뒤에 값 하나 추가** (메서드)

````python
# 💻 예제 — 개수와 추가
users = ["kim01", "lee02"]
print(len(users))       # 2
users.append("park03")  # 뒤에 추가
print(users)
print(len(users))       # 3
````

**✅ 성공 출력**
```
2
['kim01', 'lee02', 'park03']
3
```

## 2. 딕셔너리 (dict) — '이름표'로 꺼내기

- 리스트는 "몇 번째"로 꺼냅니다. 그런데 로그 한 줄은 *user·event·ip* 처럼 **이름 있는 정보**입니다.
- **딕셔너리**는 **이름표(키)-값** 쌍으로 담고, **이름으로** 꺼냅니다. 중괄호 `{ }` 를 씁니다.

> **왜 딕셔너리를 쓸까?** `log["user"]` 가 `log[0]` 보다 "무엇을 꺼내는지" 훨씬 분명하기 때문입니다.

````python
# 💻 예제 — 딕셔너리 만들고 이름으로 꺼내기
log = {"user": "kim01", "event": "login_failed"}
print(log["user"])    # "user" 이름표의 값
print(log["event"])   # "event" 이름표의 값
````

**✅ 성공 출력**
```
kim01
login_failed
```

### 📖 키가 여러 개일 때

- 딕셔너리에는 이름표-값 쌍을 여러 개 담을 수 있습니다.
- 보안 로그 한 줄을 통째로 딕셔너리로 표현하면 다루기 편합니다.

````python
# 💻 예제 — 로그 한 줄을 통째로
log = {
    "customer": "hanbit",
    "user": "kim01",
    "event": "login_failed",
    "ip": "203.0.113.5",
}
print(log["customer"])
print(log["ip"])
````

### 📖 없는 키도 안전하게 — `.get()`

- 대괄호 `log["없는키"]` 로 **없는 이름표**를 꺼내면 **에러가 나서 멈춥니다.**
- `log.get("없는키")` 는 없으면 조용히 `None`("값 없음")을 돌려주어 멈추지 않습니다.

````python
# 💻 예제 — 대괄호 vs .get()
log = {"user": "kim01"}
print(log.get("user"))     # 있으면 값 → kim01
print(log.get("country"))  # 없으면 None
````

**✅ 성공 출력**
```
kim01
None
```

### 📖 딕셔너리 훑어보기 — keys / values / items
- `keys()` 이름표만 / `values()` 값만 / `items()` 이름표+값 쌍 (모두 메서드)

````python
# 💻 (심화) keys / values
log = {"user": "kim01", "event": "login_failed"}
print(list(log.keys()))     # 이름표들
print(list(log.values()))   # 값들
````

**✅ 성공 출력**
```
['user', 'event']
['kim01', 'login_failed']
```

````python
# ✍️ (심화·출력 예측) items() 로 반복하면 무엇이 나올지 예측한 뒤 실행하세요.
log = {"user": "kim01", "event": "login_failed"}
for key, value in log.items():
    print(key, value)     # 예측: ?
````

### 📖 리스트 + 딕셔너리 조합 (실무에서 가장 많이 씀)
- 로그 **여러 줄** = "딕셔너리(로그 한 줄)를 리스트에 담기".
- 이 구조가 앞으로 로그를 훑고 거르는 실습의 핵심 재료가 됩니다.

````python
# 💻 (심화) 딕셔너리들의 리스트
logs = [
    {"user": "kim01", "event": "login_failed"},
    {"user": "lee02", "event": "login_success"},
]
print(logs[0]["user"])    # 첫 번째 로그의 user
print(logs[1]["event"])   # 두 번째 로그의 event
````

**✅ 성공 출력**
```
kim01
login_success
```

## 오늘 요약

1. **리스트 `[ ]`** — 여러 값을 순서대로. 번호(0부터)로 꺼내고, `len()` 개수, `append()` 추가.
2. **함수/메서드** — `len(값)` 은 함수, `값.append(...)` 은 메서드(값에 붙는 함수).
3. **딕셔너리 `{ }`** — 이름표(키)-값. 이름으로 꺼냄. 보안 로그 한 줄에 적합.
4. **`.get()`** — 없는 키도 에러 없이 `None`. (심화) **딕셔너리들의 리스트** = 로그 여러 줄.
