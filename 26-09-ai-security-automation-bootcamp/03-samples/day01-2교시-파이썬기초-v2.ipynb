## 용어 풀이

오늘 처음 나오는 말을 미리 정리합니다.

| 용어 | 쉽게 말하면 |
|------|-------------|
| **변수(variable)** | 값에 붙이는 이름표. `x = 7` 은 "값 7에 x라는 이름표를 붙임" |
| **자료형(type)** | 값의 종류. 숫자인지 글자인지 참/거짓인지 |
| **정수 `int`** | 소수점 없는 숫자. 예: `7`, `-3`, `0` |
| **실수 `float`** | 소수점 있는 숫자. 예: `3.14`, `0.5` |
| **문자열 `str`** | 글자(따옴표로 감쌈). 예: `"kim01"` |

| **f-string** | 문장 안에 값을 끼워 넣는 출력 방법. `f"fail {n}"` |
| **리스트 `list`** | 여러 값을 순서대로 담는 상자 `[...]` |
| **딕셔너리 `dict`** | 이름표-값 쌍으로 담는 상자 `{키: 값}` |
| **형변환** | 값의 종류 바꾸기. 글자 `"7"` → 숫자 `7` |

## 1. 변수와 기본 자료형

### 📖 변수는 값에 붙이는 '이름표'입니다

프로그램은 값을 계속 기억해야 합니다. 그래서 값에 이름표를 붙입니다. 이것이 변수입니다.
`이름 = 값` 형태로 쓰면, 그 이름으로 값을 다시 꺼내 쓸 수 있습니다.
(택배 상자에 이름표를 붙여 두면 그 이름으로 상자를 찾는 것과 같습니다.)

> **왜 한 줄에 하나씩 나눠 쓸까?** 한 줄에 변수 하나면, 값이 틀렸을 때 어느 줄이
> 문제인지 바로 보입니다. 여러 개를 붙여 쓰면 나중에 디버깅이 어려워집니다.

````python
# 💻 따라치기 (1) — 변수 3개 만들기. 값과 코드는 영어로, 설명은 주석(한국어)으로.
customer_name = "acme"              # 문자열(str): 고객사 이름
failed_login_count = 7              # 정수(int): 로그인 실패 횟수
is_alert = failed_login_count > 5   # 참거짓(bool): 7 > 5 는 참이므로 True

# print() 는 () 안의 값을 화면에 보여주는 명령입니다.
print(customer_name)
print(failed_login_count)
print(is_alert)
````

**✅ 이렇게 나오면 성공입니다**
```
acme
7
True
```
`failed_login_count > 5` 는 "7이 5보다 큰가?"라는 질문이며, 참이므로 `True`가 담깁니다.

### 📖 값의 종류 4가지와 `type()`

값에는 종류(자료형)가 있습니다. `type(값)` 을 쓰면 그 종류를 알려줍니다.

> **왜 파이썬은 자료형을 미리 안 적을까?** 어떤 언어는 `int x = 7` 처럼 종류를 먼저
> 선언합니다. 파이썬은 값을 넣는 순간 종류를 **자동 판단**합니다(동적 타이핑). 그래서
> 초보자가 시작하기 쉽습니다. (이 내용은 아래 면접 질문의 답이 됩니다.)

````python
# 💻 따라치기 (2) — 자료형별로 변수를 다양하게 만들어 봅니다.
service_name = "auth-server"   # str  : 글자
port = 8080                    # int  : 정수
cpu_usage = 73.5               # float: 소수점 숫자
is_online = True               # bool : 참/거짓

print(service_name, port, cpu_usage, is_online)
````

**✅ 이렇게 나오면 성공입니다**
```
auth-server 8080 73.5 True
```
`print()` 에 값을 쉼표로 나열하면 한 줄에 띄어쓰기로 이어서 출력됩니다.

````python
# 💻 따라치기 (3) — 위 변수들의 '종류'를 확인해 봅니다.
print(type(service_name))   # str
print(type(port))           # int
print(type(cpu_usage))      # float
print(type(is_online))      # bool
````

**✅ 이렇게 나오면 성공입니다**
```
<class 'str'>
<class 'int'>
<class 'float'>
<class 'bool'>
```
`<class 'int'>` 은 "이 값은 int(정수) 종류"라는 뜻입니다.

````python
# ✍️ 직접 (1) — 아래 값을 여러분의 값으로 바꿔 실행해 보세요.
service_name = "auth-server"   # ← 다른 서비스 이름으로
port = 8080                    # ← 다른 숫자로
print(f"service={service_name}, port={port}")
````

### 📖 f-string — 값을 문장에 끼워 넣기

값 여러 개를 한 문장으로 출력할 때 사용합니다. 문자열 앞에 **`f`** 를 붙이고,
값이 들어갈 자리에 **`{변수}`** 를 적습니다.

> **왜 f-string을 배울까?** 로그 메시지, 리포트, 알림 문구를 만들 때 계속 사용합니다.
> 오늘 확실히 익혀 두면 이후 모든 과목에서 편해집니다. 아래처럼 **쉬운 것부터 5단계**로 늘려 갑니다.

````python
# 💻 따라치기 f-string ① — 이름(값) 하나 끼우기
customer_name = "acme"
print(f"customer={customer_name}")
````

**✅ 성공 출력**
```
customer=acme
```

````python
# 💻 따라치기 f-string ② — 숫자 하나 끼우기
failed_login_count = 7
print(f"failed_login={failed_login_count}")
````

**✅ 성공 출력**
```
failed_login=7
```

````python
# 💻 따라치기 f-string ③ — 여러 값을 한 문장에
customer_name = "acme"
failed_login_count = 7
is_alert = True
print(f"[{customer_name}] failed_login={failed_login_count}, alert={is_alert}")
````

**✅ 성공 출력**
```
[acme] failed_login=7, alert=True
```
중괄호 `{}` 가 여러 개여도 각각 해당 변수 값으로 바뀝니다.

````python
# 💻 따라치기 f-string ④ — 중괄호 안에서 계산도 됩니다
success = 18
fail = 7
print(f"total={success + fail}")   # {} 안에서 덧셈 결과가 들어감
````

**✅ 성공 출력**
```
total=25
```
`{success + fail}` 처럼 중괄호 안에 계산식을 직접 넣을 수 있습니다.

````python
# 💻 따라치기 f-string ⑤ — 소수점 자리수 지정하기
cpu_usage = 73.456
print(f"cpu={cpu_usage:.1f}%")   # :.1f 는 소수점 첫째 자리까지
````

**✅ 성공 출력**
```
cpu=73.5%
```
`{값:.1f}` 는 소수점 1자리로 반올림해 보여 줍니다. 리포트에서 자주 씁니다.

````python
# ✍️ 직접 (2) — 위 5단계를 참고해, 여러분의 로그 요약 문장을 f-string으로 만들어 보세요.
user = "kim01"
fail = 3
# 아래 print 를 완성해 보세요. 예) [kim01] fail=3
print(f"")  # ← 여기 안을 채워 보세요
````

## 2. 리스트와 딕셔너리

### 📖 리스트 `[ ]` — 여러 값을 '순서대로' 담기

값이 여러 개일 때(예: 로그 여러 줄) 하나하나 변수를 만들면 번거롭습니다.
리스트는 여러 값을 순서대로 한 상자에 담습니다. 대괄호 `[ ]` 를 씁니다.

> **왜 순서(번호)로 꺼낼까?** 리스트는 "몇 번째"가 중요할 때 씁니다. 그래서 이름이 아니라
> 번호(인덱스)로 값을 꺼냅니다. 그리고 그 번호는 **0부터** 시작합니다.

````python
# 💻 따라치기 — 사용자 이름 3개를 리스트에 담기
users = ["kim01", "lee02", "park03"]   # 값들을 쉼표로 구분해 [] 안에 넣음

print(users)        # 리스트 전체
print(users[0])     # 0번째 값 (파이썬은 0부터 셈)
print(len(users))   # len() 은 개수를 세어 줌
````

**✅ 성공 출력**
```
['kim01', 'lee02', 'park03']
kim01
3
```
> **주의:** 첫 번째 값은 `users[0]`, 두 번째 값은 `users[1]` 입니다.

````python
# 💻 따라치기 — 리스트에 값 추가하기(append)
users = ["kim01", "lee02"]
users.append("park03")   # 맨 뒤에 값 하나를 추가
print(users)
print(len(users))
````

**✅ 성공 출력**
```
['kim01', 'lee02', 'park03']
3
```
`append()` 는 로그가 하나씩 들어올 때 리스트를 늘려 가는 데 자주 씁니다.

````python
# ✍️ 직접 — users 리스트에 여러분 아이디를 추가하고, 두 번째 값을 출력해 보세요.
users = ["kim01", "lee02"]
# users.append(...)  # ← 추가해 보세요
# print(...)         # ← 두 번째 값을 출력해 보세요
````

### 📖 딕셔너리 `{ }` — '이름표: 값' 쌍으로 담기

보안 로그 한 줄에는 *누가·언제·무슨 이벤트·어느 IP* 처럼 **이름이 있는 정보**가 많습니다.
이럴 때는 딕셔너리가 적합합니다. `{이름표(키): 값}` 으로 담고, **이름으로** 값을 꺼냅니다.

> **왜 리스트 대신 딕셔너리?** "몇 번째"가 아니라 "무슨 항목"이 중요할 때 씁니다.
> `log["user"]` 처럼 이름으로 꺼내면 코드가 훨씬 읽기 쉽습니다.

````python
# 💻 따라치기 — 보안 로그 한 줄을 딕셔너리로 표현하기
log_event = {
    "customer": "acme",
    "user": "kim01",
    "event": "login_failed",
    "ip": "203.0.113.5",
    "timestamp": "2026-09-01T09:12:00",
}

print(log_event["event"])   # "event" 라는 이름표의 값만 꺼내기
````

**✅ 성공 출력**
```
login_failed
```

````python
# 💻 따라치기 — 더 안전하게 꺼내기: 대괄호 vs .get()
log_event = {"user": "kim01", "event": "login_failed"}

print(log_event.get("event"))     # 있으면 값
print(log_event.get("country"))   # 없으면 None (에러 없이 넘어감)
````

**✅ 성공 출력**
```
login_failed
None
```
> **주의:** `log_event["country"]` 처럼 대괄호로 없는 이름표를 꺼내면 **에러가 나서 멈춥니다.**
> `.get()` 은 없으면 조용히 `None` 을 돌려주어 프로그램이 멈추지 않습니다.

````python
# ✍️ 직접 — 여러분의 로그 한 줄을 딕셔너리로 만들고, user 값을 꺼내 출력해 보세요.
my_log = {
    "user": "kim01",          # ← 바꿔 보세요
    "event": "login_success", # ← 바꿔 보세요
}
# print(...)  # ← my_log 의 user 값을 출력해 보세요
````

## 3. 연산자와 형변환

### 📖 연산자 — 계산하고, 비교하고, 판단하기

- **산술** `+ - * /` : 더하기·빼기·곱하기·나누기
- **비교** `> < >= <= == !=` : 크다·작다·같다·다르다 → 결과는 **참/거짓(bool)**
- **논리** `and or not` : 여러 조건을 묶기

> **왜 `==` 는 등호가 두 개일까?** `=` 는 "값을 넣어라(대입)", `==` 는 "같은지 물어봐라(비교)"
> 로 뜻이 완전히 다릅니다. 초보자가 가장 많이 헷갈리는 지점입니다.

````python
# 💻 따라치기 — 세 종류 연산자 맛보기
print(7 + 3)             # 산술: 10
print(7 > 5)             # 비교: True
print(7 > 5 and 3 > 10)  # 논리: True and False → False
````

**✅ 성공 출력**
```
10
True
False
```
`and` 는 **둘 다 참**이어야 참입니다. `3 > 10` 이 거짓이라 전체가 `False` 입니다.

````python
# 💻 따라치기 — 비교 연산은 항상 참/거짓(bool)을 돌려줍니다
count = 7
print(count == 7)   # 같은가? True
print(count != 7)   # 다른가? False
print(count >= 10)  # 10 이상인가? False
````

**✅ 성공 출력**
```
True
False
False
```
이 참/거짓 값은 3교시에서 배울 `if` 조건문의 재료가 됩니다.

### 📖 형변환 — 글자와 숫자 바꾸기 (실수 주의 구간)

파일(CSV)에서 숫자를 읽어오면 사실은 **숫자처럼 생긴 '글자'** 로 들어옵니다.
글자끼리는 더하기가 안 되므로, `int()` 로 **진짜 숫자로 바꿔** 주어야 계산됩니다.

- `int("7")` → 숫자 `7` · `str(7)` → 글자 `"7"` · `float("3.14")` → 소수 `3.14`

> **왜 이게 중요할까?** 실무 오류의 상당수가 "글자 숫자를 진짜 숫자로 안 바꿔서" 생깁니다.
> 오늘 이 감각을 잡아 두면 이후 파일·API 데이터를 다룰 때 실수를 크게 줄일 수 있습니다.

````python
# 💻 따라치기 — 글자를 숫자로 바꿔 계산하기
raw_value = "7"          # 파일에서 읽어온 값은 이렇게 '글자'인 경우가 많음
count = int(raw_value)   # int() 로 진짜 숫자로 바꿈
print(count + 1)         # 이제 계산이 됨
````

**✅ 성공 출력**
```
8
```
> **주의:** `int()` 없이 `"7" + 1` 을 하면 글자와 숫자를 더할 수 없어 **에러**가 납니다.
> 그래서 형변환이 필요합니다.

````python
# ✍️ 직접 — raw_value 를 다른 글자 숫자로 바꿔 계산해 보세요.
#          (도전) "7건" 처럼 글자가 섞인 값으로 바꾸면 무슨 일이 나는지 실험해 보세요.
raw_value = "7"          # ← "20" 등으로 바꿔 보고, 나중에 "7건" 으로도 실험
count = int(raw_value)
print(count + 1)
````

- `"7"` 은 **문자열(str)**, `7` 은 **정수(int)** 로 종류가 다릅니다. 겉모습은 같아 보여도 컴퓨터에게는 전혀 다른 값입니다.
- 문자열은 계산이 아니라 **글자 이어붙이기**로 동작합니다. 그래서 `"7" + 1` 처럼 글자와 숫자를 더하려 하면 **오류(TypeError)** 가 납니다.
- 해결법은 `int("7")` 로 **먼저 진짜 숫자로 바꾼 뒤** 계산하는 것입니다. 파일이나 API에서 읽은 숫자는 대부분 문자열이라, 이 형변환 습관이 중요합니다.
