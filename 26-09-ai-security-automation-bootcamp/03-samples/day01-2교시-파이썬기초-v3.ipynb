| 용어 | 쉽게 말하면 |
|------|-------------|

## 1. 변수와 자료형

### 1-1. 변수는 값에 붙이는 '이름표'입니다

프로그램은 값을 계속 기억해야 합니다. 그래서 값에 이름표를 붙입니다. 이것이 변수입니다.
`이름 = 값` 으로 쓰면, 그 이름으로 값을 다시 꺼내 씁니다.

> **왜 `=` 가 "같다"가 아닐까?** 파이썬에서 `=` 는 "**값을 넣어라(대입)**"라는 뜻입니다.
> "같은지 비교"는 `==`(등호 두 개)로 따로 씁니다. 이 둘을 섞으면 초보자가 가장 많이 헷갈립니다.

> **왜 한 줄에 변수 하나씩?** 값이 틀렸을 때 어느 줄이 문제인지 바로 보이기 때문입니다.

````python
# 💻 예제 — 변수를 만들고, 값을 바꾸기(재할당)
customer_name = "hanbit"   # "hanbit" 값에 customer_name 이라는 이름표를 붙임
print(customer_name)       # hanbit

customer_name = "daum"     # 같은 변수에 새 값을 넣으면 값이 바뀜(재할당)
print(customer_name)       # daum
````

**✅ 성공 출력**
```

```

### 1-2. 변수 이름 짓는 법

이름에는 **규칙(어기면 에러)** 과 **관습(어겨도 실행은 되지만 지키는 게 좋음)** 이 있습니다.

**규칙 — 어기면 에러가 납니다**
- 영문·숫자·밑줄(`_`)만 사용, **숫자로 시작 금지** (`2fail` ✗)
- 공백·하이픈 금지 (`fail count` ✗, `fail-count` ✗)
- 예약어 금지 (`if`, `for`, `class` 등은 이름으로 못 씀)

**관습 — 팀·실무의 약속**
- 파이썬 표준은 **`snake_case`** (`failed_login_count`). 참고로 다른 언어는 `camelCase`,
  `PascalCase`도 쓰지만, **파이썬에선 snake_case** 를 씁니다.
- **`is_` / `has_` / `can_`** 로 시작하면 보통 **참거짓(bool)** 값을 뜻합니다.
- **복수형**(`users`, `logs`)은 보통 **여러 개(리스트)** 를 뜻합니다.
- **안 바뀌는 값(상수)** 은 **대문자**로 (`MAX_RETRY = 5`).

> **왜 이름 규칙이 중요할까?** 이름만 봐도 "이건 참거짓이구나, 이건 목록이구나"를 알 수 있어
> 코드를 훨씬 빨리 읽고, 실수를 줄입니다.

````python
# 💻 예제 — 관습을 지킨 좋은 이름들
failed_login_count = 7                 # snake_case + 의미 있는 이름
is_blocked = False                     # is_ → 참거짓(bool)
has_permission = True                  # has_ → 참거짓(bool)
users = ["kim01", "lee02"]             # 복수형 → 여러 개(리스트)
MAX_RETRY = 5                          # 대문자 → 상수(안 바뀌는 값)

print(failed_login_count, is_blocked, has_permission, users, MAX_RETRY)
````

**✅ 성공 출력**
```
7 False True ['kim01', 'lee02'] 5
```
> **주의(어기면 에러):** `2fail = 1`(숫자 시작), `fail-count = 1`(하이픈), `class = 1`(예약어)은
> 모두 에러입니다. `camelCase`는 에러는 아니지만 파이썬 관습이 아닙니다.

### 1-3. 자료형 5가지와 `type()`

값에는 종류(자료형)가 있습니다. `type(값)` 으로 확인합니다.

> **왜 파이썬은 자료형을 미리 안 적을까?** 어떤 언어는 `int x = 7` 처럼 종류를 먼저 선언합니다.
> 파이썬은 값을 넣는 순간 종류를 **자동으로 판단**합니다(동적 타이핑). 그래서 시작이 쉽습니다.

````python
# 💻 예제 — 자료형 5가지를 모두 만들어 봅니다 (보안 관제 상황)
failed_login_count = 7          # int   : 정수
cpu_usage = 73.5                # float : 소수점 숫자
user = "kim01"                  # str   : 문자열
is_alert = True                 # bool  : 참거짓
assigned_analyst = None         # None  : 담당자 아직 미배정("값 없음")

print(failed_login_count, cpu_usage, user, is_alert, assigned_analyst)
````

**✅ 성공 출력**

### 1-4. f-string — 값을 문장에 끼워 넣기

> **왜 f-string을 배울까?** 로그 메시지·리포트·알림 문구에 계속 씁니다. 쉬운 것부터 5단계로 늘려 봅니다.

````python
# 💻 예제 ① — 값 하나 끼우기
customer_name = "hanbit"
print(f"customer={customer_name}")
````


````python
# 💻 예제 ② — 숫자 끼우기
failed_login_count = 7
print(f"failed_login={failed_login_count}")
````

**✅ 성공 출력**


````python
# 💻 예제 ③ — 여러 값을 한 문장에
customer_name = "hanbit"
failed_login_count = 7
is_alert = True
print(f"[{customer_name}] failed_login={failed_login_count}, alert={is_alert}")
````

### 2-2. 딕셔너리 `{ }` — '이름표: 값' 쌍으로 담기

보안 로그 한 줄은 *user·event·ip* 처럼 **이름 있는 정보**의 묶음이라 딕셔너리가 적합합니다.

> **왜 리스트 대신 딕셔너리?** "몇 번째"가 아니라 "무슨 항목"이 중요할 때 씁니다.
> `log["user"]` 가 `log[0]` 보다 뜻이 분명합니다.

> **집합(set)·튜플(tuple)** 은 오늘은 다루지 않습니다. 집합은 뒤(반복문 뒤 '중복 제거·집계')에서 배웁니다.

````python
# 💻 예제 — 로그 한 줄을 딕셔너리로 + 안전하게 꺼내기(.get)
log_event = {"customer": "hanbit", "user": "kim01", "event": "login_failed"}
print(log_event["event"])       # 이름으로 값 꺼내기 → login_failed
print(log_event.get("country")) # 없는 이름 → None (에러 없이 넘어감)
````

**✅ 성공 출력**
```

None
```
> **주의:** `log_event["country"]`(대괄호)로 없는 이름을 꺼내면 **에러로 멈춥니다.**
> `.get()` 은 없으면 `None` 을 돌려주어 멈추지 않습니다. (1-3의 None 과 연결됩니다.)

## 3. 연산자와 형변환

### 3-1. 연산자 — 계산·비교·판단

- **산술** `+ - * /`
- **비교** `> < >= <= == !=` → 결과는 참거짓(bool)
- **논리** `and or not`

> **왜 `==` 는 등호가 두 개일까?** `=` 는 "값을 넣어라(대입)", `==` 는 "같은지 물어봐라(비교)"
> 로 뜻이 완전히 다릅니다.

````python
# 💻 예제 — 세 종류 연산자
print(7 + 3)             # 산술 → 10
print(7 == 7)            # 비교 → True
print(7 > 5 and 3 > 10)  # 논리 → True and False → False
````

**✅ 성공 출력**
```
10
True
False
```
`and` 는 둘 다 참이어야 참입니다.

````python
# ✍️ 문제 — 예제를 참고해 아래 결과를 예상하고, 실행해 확인해 보세요.
#   목표: 각 줄이 무엇을 출력할지 먼저 종이에 적어 본 뒤 실행
count = 7
print(count != 7)        # 다른가?
print(count >= 5)        # 5 이상인가?
print(count == 5 or count > 0)  # 둘 중 하나라도 참인가?
````
